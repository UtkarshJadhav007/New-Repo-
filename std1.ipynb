```python
# Create an 10X10 array with random values and find its minimum, maximum, mean, and standard deviation 

import numpy as np 
# i) 10x10 random array 
a = np.random.rand(10, 10)
print("Minimum:", a.min())
print("Maximum:", a.max())
print("Mean:", a.mean())
print("Standard Deviation:", a.std())

# ii) Matrix multiplication
A = np.random.rand(5, 3)
B = np.random.rand(3, 2)
print("Result:")
print(np.dot(A, B))
```
