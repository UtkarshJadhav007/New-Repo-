```python
import matplotlib.pyplot as plt 
from sklearn.linear_model import LogisticRegression

# X hours Studied 
X = [[1], [2], [3], [4], [5], [6]]

# Y = Result (0 = Fail, 1 = Pass)
Y = [0,0,0,1,1,1]

# Create Model 
model = LogisticRegression()

# Trian the model
model.fit(X,Y)

# Predict result for 5 hours 
print("Prediction ", model.predict([[5]])[0])

# Plot the data 
plt.scatter(X[:3], Y[:3], color="red", label="Fail")
plt.scatter(X[3:], Y[3:], color="green", label="Pass")

# Decision boundary 
plt.axvline(x = 3.5, color="blue", linestyle="--", label="Decision Boundary")

plt.xlabel("Hours Studied")
plt.ylabel("Result")
plt.title("Logistic Regression")
plt.legend()

plt.show()

```
