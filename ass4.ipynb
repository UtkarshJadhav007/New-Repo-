```python
from sklearn.ensemble import RandomForestClassifier

# Data 
X = [[15], [30], [45], [60], [75], [90]]
Y = [0,0,0,1,1,1]

# Create Model 
model = RandomForestClassifier(n_estimators=10, random_state=42)

# Train the Model 
model.fit(X,Y)

# Predict 
print("Prediction:", model.predict([[40]]))
print("Prediction:", model.predict([[55]]))
```
